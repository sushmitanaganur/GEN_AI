```python
import string 

with open("data.txt","r") as file:
    data = file.read()

new_data=" "
for char in data:
    if char not in string.punctuation:
        new_data+=char

with open("data.txt","w") as file:
    file.write(new_data)

print("Punctuation removed successfully")
```
